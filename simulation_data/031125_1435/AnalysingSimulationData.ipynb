```python
import numpy as np
import pandas as pd
import pyarrow

def summarize_arrow(filename, ns_value):
    """Compute steady-state RNA, Protein, Ratio, and growth rate λ from one .arrow file."""
    with pyarrow.memory_map(filename, "r") as source:
        df = pyarrow.ipc.open_file(source).read_all().to_pandas()

    time = df["timestamp"].values
    RNA  = df["mq(t)"].values + df["mr(t)"].values      # total mRNA proxy
    PROT = df["r(t)"].values                            # ribosomes = protein proxy

    # --- steady-state window (last 20%) ---
    n = len(time)
    k0 = int(max(3, np.floor(0.8 * n)))
    RNA   = np.mean(RNA[k0:])
    Protein  = np.mean(PROT[k0:])
    Ratio = RNA / Protein if Protein != 0 else np.nan

    # --- instantaneous growth rate λ = d(log(a))/dt ---
    if "a(t)" in df.columns:
        signal = df["a(t)"].values
    elif "em(t)" in df.columns:
        signal = df["em(t)"].values
    else:
        signal = PROT
    λ = np.gradient(np.log(signal)) / np.gradient(time)
    λ_ss = np.mean(λ[k0:])

    return dict(ns=ns_value, RNA=RNA, Protein=Protein, Ratio=Ratio, lambda_ss=λ_ss)

import numpy as np
import pandas as pd
import pyarrow

def compute_lambda(df, time_col="timestamp"):
    """Estimate steady-state growth rate λ from log(signal) vs time."""
    time = np.asarray(df[time_col], dtype=float)

    # Try candidates in priority order
    for col in ["em(t)", "a(t)", "r(t)"]:
        if col in df.columns:
            signal = np.asarray(df[col], dtype=float)
            break
    else:
        return np.nan  # no valid growth signal found

    # Keep only positive, finite values
    mask = np.isfinite(signal) & (signal > 0)
    t, y = time[mask], signal[mask]
    if len(t) < 10:
        return np.nan

    # Use last 20% for steady state
    n = len(t)
    k0 = int(max(5, np.floor(0.8 * n)))
    t_ss = t[k0:]
    y_ss = np.log(y[k0:])

    # Perform linear regression
    if len(t_ss) < 5:
        return np.nan
    A = np.vstack([np.ones_like(t_ss), t_ss]).T
    β, _, _, _ = np.linalg.lstsq(A, y_ss, rcond=None)
    λ = β[1]  # slope
    return λ


def summarize_arrow(filename, ns_value):
    with pyarrow.memory_map(filename, "r") as source:
        df = pyarrow.ipc.open_file(source).read_all().to_pandas()

    RNA  = df["mq(t)"].values + df["mr(t)"].values
    Protein = df["r(t)"].values
    time = df["timestamp"].values

    n = len(time)
    k0 = int(max(3, np.floor(0.8 * n)))
    RNA   = np.mean(RNA[k0:])
    Protein  = np.mean(Protein[k0:])
    Ratio = RNA / Protein if Protein != 0 else np.nan

    λ_ss = compute_lambda(df)

    return dict(ns=ns_value, RNA=RNA, Protein=Protein, Ratio=Ratio, lambda_ss=λ_ss)

```

```python
summaries = []
summaries.append(summarize_arrow("seed1_ns0.7481584182562523_abx0.0.arrow", ns_value=0.748))
summaries.append(summarize_arrow("seed2_ns1.7623942844141665_abx0.0.arrow", ns_value=1.762))
summaries.append(summarize_arrow("seed3_ns0.45586113648724746_abx0.0.arrow", ns_value=0.456))
summaries.append(summarize_arrow("seed4_ns1.0365785284612912_abx0.0.arrow", ns_value=1.037))

summary_df = pd.DataFrame(summaries)
summary_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,4))
plt.scatter(summary_df["Ratio"], summary_df["lambda_ss"], color="tab:blue")
plt.xlabel("RNA/Protein ratio (φR proxy)")
plt.ylabel("Growth rate λ (1/h)")
plt.title("Growth law from simulations")
plt.grid(alpha=0.3)
plt.tight_layout()
```

```python
m, b = np.polyfit(summary_df["Ratio"], summary_df["lambda_ss"], 1)
x = np.linspace(summary_df["Ratio"].min(), summary_df["Ratio"].max(), 100)
plt.plot(x, m*x + b, "k--", label=f"λ = {m:.3f} φR + {b:.3f}")
plt.legend()
```
